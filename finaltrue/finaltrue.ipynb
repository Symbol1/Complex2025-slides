## Log Branching

```sage
r1 = x * (30 + cos(x*50)) / 10
r2 = x * (30 + cos(x*70)) / 10
freq = 3*pi * x
X = r1 * cos(freq)
X = X - X(x=-4)
Y = abs(r2) * sin(freq)
parametric_plot([X, Y], [x, -4, 10]) + points([(100, 3)], xmin=-5, xmax=25, ymin=-15, ymax=15, color='red', figsize=[5, 5])
```

## Weierstrass Function

```sage
var('w z')
(1 / (z - w)^2 - 1 / w^2).factor()
```

```sage
def wp(z, r=100):
    s = 0
    z = CDF(z)
    for a in range(-r, r):
        for b in range(-r, r):
            if a == 0 and b == 0: continue
            w = CDF(a + b*i)
            s += (2*w + z)*z / (z - w)^2 / w^2
    return 1 / z^2 + s
```

```sage
wp(1/10 + i/10, r=1).n()
```

```sage
wp(1/10 + i/10, r=2).n()
```

```sage
wp(1/10 + i/10, r=5).n()
```
